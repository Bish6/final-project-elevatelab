```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_csv("data/inventory_profit_analysis.csv")

print("Rows and columns:", df.shape)
df.head()
```

```python
analysis_df = df[
    ["Estimated_Inventory_Days", "Gross_Profit"]
].dropna()

correlation = analysis_df["Estimated_Inventory_Days"].corr(
    analysis_df["Gross_Profit"]
)

print("Rows included:", len(analysis_df))
print("Correlation:", round(correlation, 3))
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=analysis_df,
    x="Estimated_Inventory_Days",
    y="Gross_Profit",
    alpha=0.6
)

plt.title("Estimated Inventory Days and Gross Profit")
plt.xlabel("Estimated inventory days")
plt.ylabel("Gross profit")
plt.tight_layout()
plt.show()
```

## Correlation finding

The correlation between estimated inventory days and gross profit was -0.178 across 849 store-product rows. This indicates a weak negative relationship in the data. It does not prove that higher inventory causes lower profit. Inventory days are estimates based on only 37 days of sales, and rows with no sales were excluded.
